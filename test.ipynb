```python
# 测试字符串匹配问题

list1 = ["5f", "6d", "7s"]
list2 = ['5d(10)', '5f(5)6H', '6s(2)', '6p(6)', '6d(2)3F_4D', '7s_5D']

print("第一个列表:", list1)
print("第二个列表:", list2)
print()

# 分析每个目标字符串在第二个列表中的出现情况
for target in list1:
    print(f"查找 '{target}':")
    for i, item in enumerate(list2):
        if target in item:
            print(f"  在索引 {i} 找到: '{item}'")
    print()
```
